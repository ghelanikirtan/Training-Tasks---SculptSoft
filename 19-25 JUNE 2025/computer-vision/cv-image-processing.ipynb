## Computer Vision:

- Image Processing Concepts:

```python
import cv2
import os 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

Load Images:

```python
image_filenames = os.listdir('sample-images')
print(f"Image Files: {image_filenames}")

images = []
# Load Image using open-cv
for img_file in image_filenames:
    
    image = cv2.imread(os.path.join('sample-images', img_file))
    
    # Converting BGR -> RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    images.append(image)
```

Check Shape of images:

```python
print(f"Shape of Images: (Height, Width, _)")
for image_file, image in zip(image_filenames,images):
    print(f"[image] {image_file}: {image.shape}")
```

```python
plt.figure()
plt.subplot(2,2,1)

for i, image in enumerate(images):
    plt.subplot(2,2,i+1)
    plt.imshow(image)
    plt.axis('off')
plt.show()
```

Resizing:

```python
plt.figure(figsize=(10,6))

for i, image in enumerate(images):

    image = cv2.resize(image, (600,420)) # (w, h)
    plt.subplot(2,2,i+1)
    plt.imshow(image)
    plt.axis('off')
plt.show()
```

Thresholding:

- Generally used for textual based data.

```python
plt.figure(figsize=(10,6))

for i, image in enumerate(images):

    _,image = cv2.threshold(image, 127, 300, cv2.THRESH_BINARY) # (w, h)
    plt.subplot(2,2,i+1)
    plt.imshow(image)
    plt.axis('off')
plt.show()
```

Contours

```python
plt.figure(figsize=(10,6))

# for i, image in enumerate(images):
image = images[1]
plt.subplot(2,2,1)
plt.title('Original')
plt.axis('off')
plt.imshow(image)

# Grayscaling:
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.subplot(2,2,2)
plt.title('GrayScaled')
plt.axis('off')
plt.imshow(gray)

# Canny:
edged = cv2.Canny(gray, 30,200)
plt.subplot(2,2,3)
plt.title('EDGED')
plt.axis('off')
plt.imshow(edged)
    
# Contours:
contours, heir = cv2.findContours(edged, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
plt.subplot(2,2,4)
img = image.copy()
cv2.drawContours(img, contours, -1, (0,255,0), 3)
plt.title('Contours')
plt.imshow(img)
plt.axis('off')
plt.show()
```
