## Langchain:

- Loaders & Splitters.

## Types of Loader:

- Langchain's document_loaders consists of Loaders used to load the different kind of documents, files etc...


```python
import os 
```


Constants:

```python
SAMPLES_DIR = os.path.abspath('samples')
TEXT_FILE_PATH = os.path.join(SAMPLES_DIR, 'essay.txt')
PDF_FILE_PATH = os.path.join(SAMPLES_DIR, 'my-resume.pdf')
```

### 1. TextLoader:

```python
from langchain.document_loaders import TextLoader
```

```python
loader =  TextLoader(
    file_path=TEXT_FILE_PATH,
    autodetect_encoding=True
)
loader
```

```python
documents = loader.load()
```

```python
metadata = documents[0].metadata
content = documents[0].page_content
```

```python
print(f"""METADATA:::::
{metadata}
{'-'*75}
CONTENT:::::
{content}""")
```

### 2. PyPDF Loader:

- To load the pdf files.
- I will load my resume pdf file.

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader = PyPDFLoader(
    file_path=PDF_FILE_PATH
)
```

```python
pages = loader.load()
print(f"My resume's total pages: {len(pages)}")
```

```python
for i, page in enumerate(pages):
    
    print(f"""{'-'*35}Page: {i}{'-'*35}
metadata: {page.metadata}
{'-'*80}
content: {page.page_content[:500]}......
""")
```

**Lazy Loading:**

- Only loads when needed instead of loading all the pages (memory efficient loader method)
- Basically a Generator that yeilds page by page as iterated...

```python
for page in loader.lazy_load():
    print(f"""{'-'*80}
metadata: {page.metadata}
{'-'*80}
content: {page.page_content[:500]}......
""")
```

## Splitters:

- Once we load the documents they are needed to be splitted due to context window limitations and related reasons...

- Langchain comes with various splitters which splits the large data into chunks with proper strategy.

### 1. Recursive Character Text Splitter:

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
long_txt_file_path = os.path.join(SAMPLES_DIR, 'long_text.txt')
loader = TextLoader(
    long_txt_file_path,
    autodetect_encoding=True)
docs = loader.load()

#
long_text = docs[0].page_content

print(f"Length of long_text: {len(long_text)} characters")
```

```python
r_splitter = RecursiveCharacterTextSplitter(
    chunk_size=150, # max chars in chunk
    chunk_overlap=20, # Num. of chars to overlap.
    length_function = len # func. to measure chunk sz.
)
```

```python
chunks = r_splitter.split_text(long_text)
```

```python
print(f"Original Text Length: {len(long_text)}")
print(f"Number of chunks: {len(chunks)}")
```

```python
for i, chunk in enumerate(chunks):
    print(f"""{'-'*75}
CHUNK: {i+1}
Length: {len(chunk)}
{'-'*75}""")
```

### 2. Character Text Splitter:

```python
from langchain.text_splitter import CharacterTextSplitter
```

```python
c_splitter = CharacterTextSplitter(
    separator="\n", # splitting based on chars.
    chunk_size=500,
    chunk_overlap=20,
)
```

```python
chunks = c_splitter.split_text(long_text)
print(f"Original Text Length: {len(long_text)}")
print(f"Number of chunks: {len(chunks)}")
for i, chunk in enumerate(chunks):
    print(f"""{'-'*75}
CHUNK: {i+1}
Length: {len(chunk)}
{'-'*75}""")
```

### 3. Token Text Splitter:

```python
from langchain.text_splitter import TokenTextSplitter
```

```python
t_splitter = TokenTextSplitter(
    chunk_size = 50, # max. no. of tokens...
    chunk_overlap=10
)
```

```python
chunks = t_splitter.split_text(long_text)
chunks[0]
```

```python
print(f"Original Text Length: {len(long_text)}")
print(f"Number of chunks: {len(chunks)}")
for i, chunk in enumerate(chunks):
    print(f"""{'-'*75}
CHUNK: {i+1}
Length: {len(chunk)}
{'-'*75}""")
```

### More on splitters:

```python
from langchain.text_splitter import Language
```

```python
PYTHON_CODE = """
def printer_func(text):
    print(text)
    
class Animal:
    def __init__(self, name):
        self.name = name
    
    def greet(self):
        yoo = self.name
        print(f"Hello, I am {self.name}!")
        return yoo
"""
```

```python
py_splitter = RecursiveCharacterTextSplitter.from_language(
    language = Language.PYTHON,
    chunk_size = 80,
    chunk_overlap = 10
)
```

```python
chunks = py_splitter.split_text(PYTHON_CODE)

chunks
```

```python
print(f"Original Python Code length: {len(long_text)}")
print(f"Number of python code chunks: {len(chunks)}")
for i, chunk in enumerate(chunks):
    print(f"""{'-'*75}
CHUNK: {i+1}
Length: {len(chunk)} 
------
{chunk}
{'-'*75}""")
```
