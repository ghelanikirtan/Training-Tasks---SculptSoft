# Facebook's NLLB's language translation model:

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
DEVICE = 'cuda:0' if torch.cuda.is_available() else 'cpu'
DEVICE
```

**Loading The Model From Hugging Face:**

```python
inference : str = 'facebook/nllb-200-distilled-600M'
tokenizer = AutoTokenizer.from_pretrained(inference)
model = AutoModelForSeq2SeqLM.from_pretrained(inference).to(DEVICE)
```

**Using the Model for language translation:**

```python
class Translator:
    
    def __init__(self, 
                 tokenizer: AutoTokenizer,
                 model: AutoModelForSeq2SeqLM,
                 source_language='eng_Latn',
                 target_language='hin_Deva'):
        
        self.tokenizer = tokenizer
        self.model = model 
        self.source_language = source_language
        self.target_language = target_language

    def translate(self, text: str) -> str:
        self.tokenizer.src_lang = self.source_language

        inputs = self.tokenizer(text, return_tensors='pt')
        inputs = {k: v.to(DEVICE) for k, v in inputs.items()}

        forced_bos_token_id = self.tokenizer.convert_tokens_to_ids(self.target_language)
        
        output = self.model.generate(
            **inputs,
            forced_bos_token_id=forced_bos_token_id
        )
        return self.tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# English to Hindi
eng_spanish_translator = Translator(tokenizer, model)
eng_spanish_translator.translate("Hello, what are you doing?")
```

```python
# English to Gujarati:

eng_guj_translator = Translator(tokenizer, model, target_language='guj_Gujr')
eng_guj_translator.translate("Hello, what are you doing?")
```

```python
# Eng to German:
eng_german_translator = Translator(tokenizer, model, target_language='deu_Latn')
eng_german_translator.translate("Hello, with T5, we propose reframing all NLP tasks into a unified text-to-text-format.")
```

```python
# German to English:

german_eng_translator = Translator(tokenizer, model, source_language='deu_Latn', target_language='eng_Latn')
german_eng_translator.translate('Hallo, mit T5 schlagen wir vor, alle NLP-Aufgaben in ein einheitliches Text-zu-Text-Format neu zu formulieren.')
```
