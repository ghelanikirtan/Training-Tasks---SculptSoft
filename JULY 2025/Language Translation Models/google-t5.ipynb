# Google's T5 based Language Translation:

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
DEVICE = 'cuda:0' if torch.cuda.is_available() else 'cpu'
DEVICE
```

**Loading The Model From Hugging Face:**

```python
inference : str = 'google-t5/t5-large'
tokenizer = AutoTokenizer.from_pretrained(inference)
model = AutoModelForSeq2SeqLM.from_pretrained(inference).to(DEVICE)
```

**Using the Model for language translation:**

```python
class Translator:
    
    def __init__(self, 
                 tokenizer : AutoTokenizer,
                 model : AutoModelForSeq2SeqLM,
                 base_language = 'English',
                 target_language = 'Hindi'):
        
        self.tokenizer : AutoTokenizer = tokenizer
        self.model : AutoModelForSeq2SeqLM = model 
        self.base_language = base_language
        self.target_language = target_language
        self.t5_task = f"translate {self.base_language} to {self.target_language}:" #append the input here...
        
    
    def translate(self, text: str ) -> str:
        input_ids = self.tokenizer(f"{self.t5_task} {text}", return_tensors='pt').to(DEVICE)
        output = self.model.generate(**input_ids)
        return self.tokenizer.decode(output[0], skip_special_tokens=True)
        
        
```

Different Language translators:

```python
# English to spanish
eng_spanish_translator = Translator(tokenizer, model, 'English', 'Spanish')
eng_spanish_translator.translate("Hello, Good morning!")
```

```python
# English to French:

eng_french_translator = Translator(tokenizer, model, 'English', 'French')
eng_french_translator.translate("Hello, Good Morning!")
```

```python
# English to German:

eng_german_translator = Translator(tokenizer, model, 'English', 'German')
eng_german_translator.translate("Hello, with T5, we propose reframing all NLP tasks into a unified text-to-text-format.")
```
